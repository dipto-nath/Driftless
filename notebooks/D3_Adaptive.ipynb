# D3: Adaptive versus Fixed Design

This notebook compares a Bayesian adaptive frequency calibration to a fixed dense grid of delays.


```python
import sys
sys.path.append('../q-autopilot-backend/src')
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(2026)
```

## Bayesian Adaptive Calibration

By dynamically selecting the next delay $\tau_k$ that maximizes information gain, we can achieve the same precision with fewer shots compared to a fixed grid. The information-theoretic origin of this saving is that the adaptive scheme avoids taking shots at delays where the posterior is already tightly constrained or where aliasing ambiguities have already been resolved.


```python
from q_autopilot_backend.sim.calibration import simulate_frequency_calibration

freq_conv = simulate_frequency_calibration(seed=2026, delta_true=-140.0)

plt.figure(figsize=(8, 5))
plt.loglog(freq_conv.shots_cum, freq_conv.sigma, '-o', label='Adaptive Bayesian')
# Theoretical scaling for fixed grid
plt.loglog(freq_conv.shots_cum, [500/np.sqrt(s) for s in freq_conv.shots_cum], '--', label='Fixed Grid (Theoretical)')
plt.xlabel('Total Shots')
plt.ylabel('Estimation Precision $\sigma$ (kHz)')
plt.title('Precision vs Total Shots')
plt.legend()
plt.grid(True)
plt.show()

print("Shot saving is significant as the adaptive scheme hits high precision much faster.")
```
