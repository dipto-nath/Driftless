This notebook implements the three-level transmon simulator with drift and verifies the static case.


```python
import sys
sys.path.append('../q-autopilot-backend/src')
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(2026)
```

## Hamiltonian and Drift Parameters

The transmon is modeled in a rotating frame with three levels. The anharmonicity is $\alpha/2\pi = -300$ MHz to model leakage to $|2\rangle$. Detuning $\Delta(t)$ is modeled as an Ornstein-Uhlenbeck (OU) process combined with random telegraph noise. Drive gain $g(t)$ is modeled as an OU process with a sinusoidal diurnal variation.


```python
from q_autopilot_backend.sim.transmon import simulate_populations, gate_error_oracle

# Simulate a static Gaussian pi pulse with zero detuning and unit gain
res = simulate_populations(with_drag=False, beta_ns=0.0, delta=0.0, gain=1.0)

t_ns = res['t_ns']
p0 = res['p0']
p1 = res['p1']
p2 = res['p2']

plt.figure(figsize=(8, 5))
plt.plot(t_ns, p0, label='P0 (Ground)')
plt.plot(t_ns, p1, label='P1 (Excited)')
plt.plot(t_ns, p2, label='P2 (Leakage)')
plt.xlabel('Time (ns)')
plt.ylabel('Population')
plt.title('Transmon Populations during Gaussian $\pi$ pulse')
plt.legend()
plt.grid(True)
plt.show()

print(f"Leakage into |2>: {res['leakage']:.6f}")
print(f"Gate Error ε: {res['gate_error']:.6f}")
```
